# Sample dataset: column guide

Each row is a company-year financial-statement observation. Dates are shown as `MM/DD/YY`; monetary values are reported in the source dataset's units (the file does not state the currency or scaling, so confirm this before interpreting magnitudes). Blank values denote data that were not reported or are unavailable.

| Column | Meaning |
|---|---|
| *(unnamed first column)* | Row index carried over when the CSV was exported; it is not a business variable. |
| `id` | Company identifier. |
| `stmt_date` | Closing date of the financial statement. |
| `HQ_city` | Coded city of the company's headquarters; use the relevant geographic-code lookup to translate it. |
| `legal_struct` | Legal form of the company (for example, `SRL` or `SAU`). |
| `ateco_sector` | ATECO economic-activity/industry code. |
| `def_date` | Company definition/incorporation date recorded by the source. |
| `fs_year` | Financial-statement fiscal year. |
| `asst_intang_fixed` | Intangible fixed assets. |
| `asst_tang_fixed` | Tangible fixed assets (property, plant, and equipment). |
| `asst_fixed_fin` | Financial fixed assets (long-term financial investments/receivables). |
| `asst_current` | Current assets. |
| `AR` | Accounts receivable. |
| `cash_and_equiv` | Cash and cash equivalents. |
| `asst_tot` | Total assets. |
| `eqty_tot` | Total shareholders' equity. |
| `eqty_corp_family_tot` | Total equity for the corporate-family grouping, where reported by the source. |
| `liab_lt` | Total long-term liabilities. |
| `liab_lt_emp` | Long-term employee-related liabilities (such as pension/severance obligations). |
| `debt_bank_st` | Short-term bank debt. |
| `debt_bank_lt` | Long-term bank debt. |
| `debt_fin_st` | Other short-term financial debt. |
| `debt_fin_lt` | Other long-term financial debt. |
| `AP_st` | Short-term accounts payable. |
| `AP_lt` | Long-term accounts payable. |
| `debt_st` | Total short-term debt. |
| `debt_lt` | Total long-term debt. |
| `rev_operating` | Operating revenue (revenue from the core business). |
| `COGS` | Cost of goods sold. |
| `prof_operations` | Operating profit/loss. |
| `goodwill` | Goodwill recognized on the balance sheet. |
| `inc_financing` | Financial income. |
| `exp_financing` | Financial expenses. |
| `prof_financing` | Financial profit/loss. |
| `inc_extraord` | Extraordinary/non-recurring income. |
| `taxes` | Taxes reported in the income statement. |
| `profit` | Net profit/loss for the period. |
| `days_rec` | Days receivable outstanding: an estimate of average collection time. |
| `ebitda` | Earnings before interest, taxes, depreciation, and amortization. |
| `roa` | Return on assets, typically expressed as a percentage. |
| `roe` | Return on equity, typically expressed as a percentage. |
| `wc_net` | Net working capital (current assets less current liabilities). |
| `margin_fin` | Financial margin as defined by the data provider. |
| `cf_operations` | Cash flow from operating activities. |

**Note:** `HQ_city`, `legal_struct`, and `ateco_sector` are coded categorical fields. Treat them as categories in modeling (or map them with the supplied codebooks), rather than assuming their numeric values are ordered quantities.
